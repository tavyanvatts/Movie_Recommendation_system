<h1>2. Univariate Analysis</h1>

```python
# what columns we keep----> type, director, cast, country, rating, duration, listed_in, description, show_id, release_date
```

```python
data = data[['title','type' , 'director','cast','country','rating','duration','listed_in','description','show_id','release_year']]
```

```python
data.sample(5)
```

```python
data.isnull().sum()
```

```python
data = data.dropna().reset_index(drop=True)
```

```python
data.isnull().sum()
```

```python
def convert(str):
  str = str.replace(" Movies", "")
  str = str.replace(" TV Shows", "")
  str = str.split(",")
  return str
```

```python
data['listed_in'] = data['listed_in'].apply(convert)
```

```python
data.head()
```

```python
def convertC(str):
  str = str.split(",")
  if len(str) >= 3:
    return str[:3]
  else:
    return str
```

```python
data['cast'] = data['cast'].apply(convertC)
```

```python
data['description'] = data['description'].apply(lambda x: x.split())
```

```python
data['listed_in'] = data['listed_in'].apply(lambda x: [i.replace(" ", "") for i in x])
data['cast'] = data['cast'].apply(lambda x: [i.replace(" ", "") for i in x])
data['director'] = data['director'].apply(lambda x: ["".join(x)])
```

```python
data['director'] = data['director'].apply(lambda x: [i.replace(" ", "") for i in x])
```

```python
data['tags'] = data['director'] + data['cast'] + data['listed_in'] + data['description']
```

```python
newdata = data[['title','show_id', 'type', 'country', 'rating', 'duration', 'release_year', 'tags']]
```

```python
newdata['tags'] = newdata['tags'].apply(lambda x: " ".join(x))
newdata['tags'] = newdata['tags'].apply(lambda x: x.lower())
```

```python
import nltk
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()
```

```python
def stem(text):
  y = []
  for i in text.split():
    y.append(ps.stem(i))
  return " ".join(y)
```

```python
newdata['tags'] = newdata['tags'].apply(stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(newdata['tags']).toarray()
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(vectors)
```

```python
def recommend(movie):
  movie_index = newdata[newdata['title'] == movie].index[0]
  distances = similarity[movie_index]
  movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x:x[1])[1:6]
  for i in movies_list:
    print(newdata.iloc[i[0]].title)
```
